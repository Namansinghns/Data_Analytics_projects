```python
import pandas as pandas
import numpy as np
import matplotlib.pyplot as plt
```

```python
porter = pandas.read_csv("/content/dataset.csv")
porter
```

```python
porter.duplicated().sum()
```

```python
porter.info()
```

```python
porter.isnull().sum()
```

```python
round(porter["market_id"].mean())

```

```python
porter["market_id"].fillna(round(porter["market_id"].mean()),inplace=True)
```

```python
porter["store_primary_category"].fillna("unknown",inplace=True)
```

```python
porter["order_protocol"].fillna(porter["order_protocol"].mean(),inplace=True)
```

```python
porter["total_onshift_partners"].fillna(porter["total_onshift_partners"].mean(),inplace=True)
porter["total_busy_partners"].fillna(porter["total_busy_partners"].mean(),inplace=True)
porter["total_outstanding_orders"].fillna(porter["total_outstanding_orders"].mean(),inplace=True)
```

```python
porter = porter.dropna(subset  = ["actual_delivery_time"])
```

```python
porter.isnull().sum()
```

```python
porter.info()
```

```python
porter
```

```python
porter["created_at"] = porter["created_at"].astype("datetime64[ns]")
```

```python
porter.columns = porter.columns.str.strip()
```

```python
porter["actual_delivery_time"] = porter["actual_delivery_time"].astype("datetime64[ns]")
```

```python
porter.info()
```

```python
porter.head()
```

```python
porter["Month"]  = pandas.to_datetime(porter["created_at"]).dt.month_name()
porter["Hour"] = pandas.to_datetime(porter["created_at"]).dt.hour
porter["Day"] = pandas.to_datetime(porter["created_at"]).dt.day_name()
```

```python
porter.head()
```

```python
order = porter.groupby("market_id")["total_items"].sum().reset_index(name="Total_items")   # 1
order
```

```python
plt.figure(figsize=(10,6),facecolor="orange")
plt.bar(order["market_id"],order["Total_items"])
plt.xlabel("Market ID")
plt.ylabel("Total Items")
plt.title("Total Items by Market ID")
plt.show()
```

```python
porter["delivery_time"] = porter["actual_delivery_time"] - porter["created_at"]   # 2
porter.head()
```

```python
porter["delivery_time_minute"] = porter["delivery_time"].dt.total_seconds()/60
porter.head()
```

```python
porter["delivery_time_minute"].mean()
```

```python
s = porter.groupby("store_primary_category")["delivery_time_minute"].mean().sort_values(ascending=True).head(10)
s
```

```python
plt.figure(figsize=(10,6),facecolor="darkgrey")
plt.barh(s.index,s.values)
plt.xlabel("Delivery Time")
plt.ylabel("Store Primary Category")
plt.title("Average Delivery Time by Store Primary Category")
plt.show()
```

```python
hourly_order = porter.groupby("Hour")["total_items"].sum().reset_index(name="count").sort_values(by ="count",ascending=False).head(5)  # 3 peak hours of orders
hourly_order
```

```python
plt.figure(figsize=(10,6),facecolor="lightgreen")
plt.bar(hourly_order["Hour"],hourly_order["count"])
plt.xlabel("Hour")
plt.ylabel("Total Items")
plt.title("Total Items by Hour")
plt.show()
```

```python
peak_hours = hourly_order.loc[hourly_order["count"].idxmax()]
peak_hours
```

```python
highest_days = porter.groupby("Day")["total_items"].sum().sort_values(ascending=False).reindex(["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"])
highest_days                       # 4 highest order volume by days
```

```python
plt.figure(figsize=(10,6),facecolor="lightblue")
plt.plot(highest_days,color = "Green")
plt.xlabel("Day")
plt.ylabel("Total Items")
plt.title("Total Items by Day")
plt.show()
```

```python
corr = round(porter["total_items"].corr(porter["subtotal"]),4) # 5 correlation between total items and subtotal
corr
```

```python
plt.figure(figsize=(10,6),facecolor="pink")
plt.scatter(porter["total_items"],porter["subtotal"])
plt.xlabel("Total Items")
plt.ylabel("Subtotal")
plt.title("Total Items vs Subtotal")
plt.show()
```

```python
porter["delivery_time_minute"].describe()  # 6 distribution of delivery time
```

```python
porter["delivery_time_minute"].min()
```

```python
plt.figure(figsize=(10,6),facecolor="lightblue")
plt.hist(porter["delivery_time_minute"],bins=5)
plt.xlabel("Delivery Time")
plt.ylabel("Frequency")
plt.title("Distribution of Delivery Time")
plt.show()
```

```python
plt.figure(figsize=(10,6),facecolor="lightgreen")
porter.boxplot(column="delivery_time_minute")
plt.xlabel("Delivery Time")
plt.ylabel("Frequency")
plt.title("Distribution of Delivery Time")
plt.show()
```

```python
filtered = porter.loc[porter["delivery_time_minute"] <=30]   # 7 order delivery less than 30 minute
```

```python
s = filtered.groupby("delivery_time_minute")["total_items"].value_counts().reset_index(name = "total_item").sort_values(by="delivery_time_minute")
s
```

```python
busy_parterns_corr = round(porter["total_items"].corr(porter["total_busy_partners"]),4)  # 8 correaltion between order_volume and busy partern
busy_parterns_corr
```

```python
plt.figure(figsize=(10,6),facecolor="skyblue")
plt.scatter(porter["total_items"],porter["total_busy_partners"])
plt.xlabel("Total Items")
plt.ylabel("Total Busy Partners")
plt.title("Total Items vs Total Busy Partners")
plt.show()
```

```python
average_delivery_time = porter.groupby("Day")["delivery_time_minute"].mean().reindex(["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"])
average_delivery_time        # 9  average delivery time by weekdays
```

```python
plt.figure(figsize=(10,6),facecolor="lightcoral")
plt.plot(average_delivery_time,color="red")
plt.xlabel("Day")
plt.ylabel("Delivery Time")
plt.title("Average Delivery Time by Day")
plt.show()
```

```python
s = porter.groupby("store_primary_category")["delivery_time_minute"].mean().sort_values(ascending=False).head()  # 10 average delivery time by store category
s
```

```python
plt.figure(figsize=(10,6),facecolor="lightcoral")
plt.plot(s,color="blue")
plt.xlabel("Store Primary Category")
plt.ylabel("Delivery Time")
plt.title("Average Delivery Time by Store Category")
plt.show()
```

```python
s = porter.groupby("order_protocol")["delivery_time_minute"].mean().sort_values(ascending=False)  # 11 average delivery time by order protocol
s
```

```python
plt.figure(figsize=(10,6),facecolor="lightcoral")
plt.bar(s.index,s.values,color="green")
plt.xlabel("Order Protocol")
plt.ylabel("Delivery Time")
plt.title("Average Delivery Time by Order Protocol")
plt.show()
```

```python
porter.groupby(["total_items","market_id"])["subtotal"].mean().sort_values(ascending=False)  # 12 subtotal order values by market id
```

```python
top_hours = porter.groupby("Hour").size().nlargest(5).index  # 13 corr between peak hours and onshift partner
top_hours
```

```python
porter["is_peak_hours"]  = porter["Hour"].isin(top_hours).astype(int)
```

```python
corr_onshift = round(porter["is_peak_hours"].corr(porter["total_onshift_partners"]),4)
corr_onshift
```

```python
plt.figure(figsize=(10,6),facecolor="lightgreen")
plt.scatter(porter["is_peak_hours"],porter["total_onshift_partners"])
plt.xlabel("Is Peak Hours")
plt.ylabel("Total On Shift Partners")
plt.show()
```
