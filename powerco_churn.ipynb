# conclusion

1.The dataset shows that most customers stay with the company, while only a small percentage churn.
This indicates that churn is relatively low but still important because losing customers affects revenue.

2.Customers who churn generally have lower electricity consumption compared to customers who stay.

3.The forecasted consumption is usually higher than actual consumption for churn customers, indicating possible overestimation of demand.

4.Forecast electricity prices for churned customers are slightly higher, which suggests that higher expected prices may contribute to churn.

5.Customers with lower gross margins are more likely to churn.

6.Aggregating the price dataset shows the average electricity price paid by each customer.

7.Customers who churn tend to experience slightly higher peak electricity prices compared to customers who stay.

8.The difference between peak and off-peak prices helps measure price sensitivity.

9.Most variables have weak correlation with churn, meaning churn is influenced by multiple factors rather than a single variable.

10.Low-consumption customers represent a higher churn risk group.


# Overall Business Insight

1.Key findings from the analysis:

2.Churn rate is relatively low but financially important.

3.Low consumption customers churn more frequently.

4.Higher electricity prices and price fluctuations may contribute to churn.

5.High-consumption customers generate the most revenue and should be retained.

6.Electricity prices show temporal fluctuations, indicating possible market dynamics.

# Business Recommendation

1.Provide loyalty offers or discounts for high-value customers.

2.Develop targeted retention strategies for low-consumption customers.

3.Monitor price sensitivity and volatility to reduce churn risk.
