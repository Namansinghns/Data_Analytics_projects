```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
average_2021 = world["2021"].mean()
round(average_2021,2)
```

```python
max_values = world.loc[world["2021"].idxmax(),"Country Name"]
max_values
```

```python
years = [str(year) for year in range(1998,2023)]
world_year = world[years]
```

```python
global_avg = world_year.mean()
global_avg
```

```python
global_av = global_avg.to_frame(name = "unemployment_rate")
```

```python
global_avg_index = global_avg.index.astype(int)
```

```python
plt.figure(figsize=(10,6))
plt.plot(global_avg_index,global_avg)
plt.title("Global Unemployment Rate")
plt.xlabel("Year")
plt.ylabel("Unemployment Rate")
plt.show()
```

```python
difference = world[["Country Name","2011","2021"]]
difference
```

```python
difference["reductions"] = world["2011"] - world["2021"]    # 4
```

```python
max_reduction = difference["reductions"].max()
```

```python
country = world[difference["reductions"] == max_reduction]["Country Name"].values[0]
print("country name highest reduction:", country)
print("max reduction:", max_reduction)
```

```python
years = ["2008","2009","2010"]
crises = world[years].mean()
crises
```

```python
crises.plot(marker = "o",figsize=(10,6))
plt.title("Financial_crises_impact")
plt.xlabel("year")
plt.ylabel("average")
plt.show()
```

```python
pandemic = ["2019","2020","2021"]
pandemic_avg = world[pandemic].mean()
pandemic_avg
```

```python
pandemic_avg.plot(marker = "o",figsize=(10,6))
plt.title("covid_panademic_effect")
plt.xlabel("year")
plt.ylabel("average")
plt.show()
```

```python
country_deviation = [str(year) for year in range(1998,2022)]
world["country_deviation"] = world[country_deviation].std(axis = 1)
```

```python
stable_country = world.sort_values("country_deviation").iloc[0]
```
