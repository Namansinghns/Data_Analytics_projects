```
import pandas as pd
import matplotlib.pyplot as plt
```

```
store  = pd.read_csv("/content/Retail Store Dataset.xlsx - Retail Store Data.csv")
store
```

```
store.isnull().sum()
```

```
store.info()
```

```
store[store.duplicated()]
```

```
drop_duplicate = store.drop_duplicates()
drop_duplicate
```

```
total_sales = store["Total Amount"].sum()
total_sales
```

```
average_sales = store["Total Amount"].mean()
average_sales
```

```
yearly_product = store.groupby(["Year","Product Category"])["Total Amount"].sum()
yearly_product
```

```
product_quantity = store.groupby("Product Category")["Quantity"].sum()
product_quantity
```

```
plt.figure(figsize = (5,5))
plt.bar(product_quantity.index,product_quantity.values)
plt.xlabel("Product Category")
plt.ylabel("Total Quantity")
plt.title("Total Quantity by Product Category")
plt.show()
```

```
average_sales_product = store.groupby("Product Category")["Total Amount"].mean()
average_sales_product
```

```
plt.figure(figsize = (5,5))
plt.bar(average_sales_product.index,average_sales_product.values)
plt.xlabel("Product Category")
plt.ylabel("Average Sales")
plt.title("Average Sales by Product Category")
plt.show()
```

```
yearly_total_sales = store.groupby("Year")["Total Amount"].sum()
yearly_total_sales
```

```
store["Date"] = pd.to_datetime(store["Date"])
store["Day"] = store["Date"].dt.day_name()
monthly_sales = store.groupby("Day")["Total Amount"].sum().reindex(["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"])
monthly_sales
```

```
plt.figure(figsize = (10,5))
plt.plot(monthly_sales,marker = "^")
plt.xticks(rotation = 45)
plt.xlabel("Day")
plt.ylabel("Total Amount")
plt.title("Monthly sales")
plt.show()
```

```
store["Date"] = pd.to_datetime(store["Date"])
store["Month"] = store["Date"].dt.month_name()
monthly_sales = store.groupby("Month")["Total Amount"].sum().reindex(["January","February","March","April","May","June","July","August","September","October","November","December"])
monthly_sales
```

```
plt.figure(figsize = (10,5))
plt.plot(monthly_sales,marker = "^")
plt.xticks(rotation = 45)
plt.xlabel("Month")
plt.ylabel("Total Amount")
plt.title("Monthly sales")
plt.show()
```

```
store["Date"] = pd.to_datetime(store["Date"])
store["Year"] = store["Date"].dt.year
monthly_sales = store.groupby("Year")["Total Amount"].sum()
monthly_sales
```

```
plt.figure(figsize = (10,5))
plt.plot(monthly_sales,marker = "^", color = "r")
plt.xticks(rotation = 45)
plt.xlabel("Year")
plt.ylabel("Total Amount")
plt.title("Monthly sales")
plt.show()
```

```
sales_product = store.groupby("Product Category")["Total Amount"].sum()
sales_product
```

```
plt.figure(figsize = (5,5))
plt.bar(sales_product.index,sales_product.values)
plt.xlabel("Product Category")
plt.ylabel("Total Amount")
plt.title("Sales by Product Category")
plt.show()
```

```
gender_sales = store.groupby("Gender")["Total Amount"].sum()
gender_sales
```

```
plt.figure(figsize = (5,5))
plt.bar(gender_sales.index,gender_sales.values)
plt.xlabel("Gender")
plt.ylabel("Total Amount")
plt.title("Sales by Gender")
plt.show()
```

```
age_sales = store.groupby("Age")["Total Amount"].sum()
age_sales
```

```
plt.figure(figsize= ( 10, 5))
plt.bar(age_sales.index,age_sales.values)
plt.xticks(rotation = 45)
plt.xlabel("Age")
plt.ylabel("Total Amount")
plt.title("Sales by Age")
plt.show()
```

```
store["Date"] = pd.to_datetime(store["Date"])
store["Month_Year"] = store["Date"].dt.strftime("%Y-%m")
monthly_sales = store.groupby("Month_Year")["Total Amount"].sum()
monthly_sales
```

```
plt.figure(figsize = (15,7))
plt.plot(monthly_sales,marker = "^", color = "Teal")
plt.xticks(rotation = 45)
plt.xlabel("Months")
plt.ylabel("Total Amount")
plt.title("Monthly sales")
plt.tight_layout()
plt.show()
```

```
store["Month"] = store["Date"].dt.month_name()
store["Year"] = store["Date"].dt.year
monthly_sales = store.groupby(["Month","Year"])["Total Amount"].sum()
monthly_sales
```

```
compare_product = store.groupby("Product Category")["Price per Unit"].sum()
compare_product
```

```
plt.figure(figsize = (5,5))
plt.bar(compare_product.index,compare_product.values)
plt.xlabel("Product Category")
plt.ylabel("Total Amount")
plt.title("Sales by Product Category")
plt.show()
```
